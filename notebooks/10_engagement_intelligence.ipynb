# 10. Engagement Intelligence
Aggregate workforce engagement scores across departments.

```
import pandas as pd

df = pd.read_csv("../data/processed/engagement_processed.csv")
dept_eng = df.groupby("Department")["EngagementScore"].agg(["mean", "median", "count"]).round(1)
print("Department Engagement Breakdown:")
display(dept_eng.sort_values(by="mean", ascending=False))
```

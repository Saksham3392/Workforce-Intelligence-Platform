# 02. Data Validation
Define explicit assertions and validation rules to catch corrupt data early.

```
import pandas as pd

df = pd.read_csv("../data/raw/employee_attrition.csv")

# Schema & Range assertions
assert df['Age'].between(18, 100).all(), "Age out of range"
assert df['EmployeeID'].is_unique, "Duplicate EmployeeID found"
assert set(df['AttritionRisk'].unique()) <= {'Yes', 'No'}, "Unexpected Attrition value"
print("All employee attrition schema assertions passed successfully!")
```

```
df_eng = pd.read_csv("../data/raw/hr_performance_engagement.csv")
assert df_eng['EngagementScore'].between(0, 100).all(), "EngagementScore out of range"
assert df_eng['Employee ID'].is_unique, "Duplicate Employee ID found in performance dataset"
print("All engagement schema assertions passed successfully!")
```

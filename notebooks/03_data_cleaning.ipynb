# 03. Data Cleaning
Clean raw datasets, impute missing values, standardize strings, and save to `data/processed/`.

```
import os
import pandas as pd

df_attrition = pd.read_csv("../data/raw/employee_attrition.csv")
df_eng = pd.read_csv("../data/raw/hr_performance_engagement.csv")

# Standardize column types and clean strings
df_attrition["Department"] = df_attrition["Department"].str.strip()
df_attrition["JobRole"] = df_attrition["JobRole"].str.strip()

os.makedirs("../data/processed", exist_ok=True)
df_attrition.to_csv("../data/processed/employee_attrition_processed.csv", index=False)
df_eng.to_csv("../data/processed/engagement_processed.csv", index=False)
print("Cleaned datasets written to data/processed/")
```

# 05. Feature Engineering
Construct model features reflecting work-life stress, compensation progression, and stagnation.

```
import pandas as pd
import numpy as np

df = pd.read_csv("../data/processed/employee_attrition_processed.csv")

current_year = 2025
df["Promotion_Gap"] = current_year - df["LastPromotionYear"]
df["Salary_Per_Year_At_Company"] = df["MonthlySalary"] / (df["YearsAtCompany"] + 1)
df["Overtime_Ratio"] = df["OvertimeHoursPerMonth"] / 40.0
df["Tenure_Ratio"] = df["YearsAtCompany"] / (df["Age"] - 18 + 1)
df["Burnout_Index"] = (df["Overtime_Ratio"] * 0.5) + np.clip((5.0 - df["WorkLifeBalanceScore"]) / 5.0, 0, 1) * 0.5

df[["EmployeeID", "Promotion_Gap", "Salary_Per_Year_At_Company", "Burnout_Index"]].head()
```

# 09. Model Versioning
Structured model artifact storage and metadata management.

```
import os
import joblib
import json

model_path = "../models/v1/attrition_pipeline.joblib"
meta_path = "../models/v1/metadata.json"

assert os.path.exists(model_path), "Model pipeline file missing!"
assert os.path.exists(meta_path), "Metadata JSON missing!"

pipeline = joblib.load(model_path)
with open(meta_path) as f:
    meta = json.load(f)

print(f"Loaded {meta['model_name']} (Version: {meta['version']}) successfully.")
print(f"Algorithm: {meta['algorithm']}, ROC-AUC: {meta['metrics']['roc_auc']}")
```

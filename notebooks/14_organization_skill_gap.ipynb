# 14. Organization-Wide Skill Gap Rollup
Aggregate missing skills across all employees and assign severity.

```
import pandas as pd
from collections import Counter

df_emp = pd.read_csv("../data/processed/employee_intelligence.csv")

all_gaps = []
for gaps in df_emp["Skill_Gap"].dropna():
    if gaps != "None (Proficient)":
        for s in gaps.split(", "):
            all_gaps.append(s)

gap_counts = pd.Series(Counter(all_gaps)).sort_values(ascending=False).to_frame("EmployeesMissing")

def get_severity(count):
    if count >= 100:
        return "HIGH"
    elif count >= 50:
        return "MEDIUM"
    return "LOW"

gap_counts["Severity"] = gap_counts["EmployeesMissing"].apply(get_severity)
print("Critical Organization Skill Gaps:")
display(gap_counts.head(15))
```

# 04. Data Relationships
Inspect joins and relationship integrity across employee, engagement, and role-skills tables.

```
import pandas as pd

df_emp = pd.read_csv("../data/processed/employee_attrition_processed.csv")
df_eng = pd.read_csv("../data/processed/engagement_processed.csv")
df_skills = pd.read_csv("../data/processed/role_skills.csv")

# Verify 1:1 join on EmployeeID
merged = pd.merge(df_emp, df_eng, left_on="EmployeeID", right_on="Employee ID", how="inner")
print("Merged Shape (1:1 join):", merged.shape)
assert len(merged) == len(df_emp), "Loss of records during join!"
print("Data relationship join verified!")
```

# 17. O*NET Skills & Workforce Intelligence
Comprehensive analysis, data processing, and querying of official O*NET Essential Skills and Software Technologies.

```python
import pandas as pd
import numpy as np

# Load processed O*NET datasets
df_ess = pd.read_csv('../data/processed/onet_essential_skills_processed.csv')
df_soft = pd.read_csv('../data/processed/onet_software_skills_processed.csv')
df_hot = pd.read_csv('../data/processed/onet_hot_technologies.csv')
df_matrix = pd.read_csv('../data/processed/onet_role_skills_matrix.csv')

print(f'Total Essential Skills records: {len(df_ess)}')
print(f'Total Software Skills records: {len(df_soft)}')
print(f'Unique O*NET Occupations: {df_ess["O*NET-SOC Code"].nunique()}')
print(f'Unique Software Tools: {df_soft["SoftwareName"].nunique()}')
```

## 1. Essential Skills Analysis
Distribution of Importance (Scale 1-5), Level (Scale 0-7), and Composite SkillScore across occupations.

```python
print('Essential Skills Summary Statistics:')
display(df_ess[['Importance', 'Level', 'SkillScore']].describe())

print('\nTop Most Critical Skills Across All Occupations (by Mean Importance):')
display(df_ess.groupby('Element Name')[['Importance', 'Level', 'SkillScore']].mean().sort_values(by='Importance', ascending=False))
```

## 2. Workplace Software & Hot Technologies Analysis
Identification of Hot Technologies and In-Demand Software across the modern labor market.

```python
print('Top 15 Hot Technologies Across All Occupations:')
display(df_hot.head(15))

print('\nTop Software Categories:')
display(df_soft['CategoryName'].value_counts().head(10).to_frame('ToolCount'))
```

## 3. Role Benchmarking Matrix
Mapping company internal job roles to standard O*NET titles, top essential skills, and in-demand software tools.

```python
print('O*NET Role Intelligence Benchmarks:')
display(df_matrix)
```
